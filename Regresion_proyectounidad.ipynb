```python
import urllib.request

url = "https://archive.ics.uci.edu/static/public/733/water%2Bquality%2Bprediction-1.zip"

archivo_zip = "water_quality_prediction.zip"

urllib.request.urlretrieve(url, archivo_zip)

print("Descarga finalizada:", archivo_zip)
```

```python
import zipfile

with zipfile.ZipFile("water_quality_prediction.zip", "r") as archivo:
    archivo.extractall("datos_agua")

# Mostrar los archivos extraídos
import os

for carpeta, subcarpetas, archivos in os.walk("datos_agua"):
    for nombre in archivos:
        print(os.path.join(carpeta, nombre))
```

```python
from scipy.io import loadmat

datos_mat = loadmat("datos_agua/water_dataset.mat")

# Mostrar las claves disponibles
print(datos_mat.keys())
```

```python
from scipy.io import loadmat

datos_mat = loadmat("datos_agua/water_dataset.mat")

# Mostrar las claves disponibles
print(datos_mat.keys())
```

```python
import numpy as np

for clave in datos_mat:
    if not clave.startswith("__"):
        objeto = datos_mat[clave]

        print(f"\nVariable: {clave}")
        print("Tipo:", type(objeto))
        print("Dimensiones:", np.shape(objeto))

        if clave in ["features", "location_ids", "location_group"]:
            print("Contenido:\n", objeto)
```

```python
caracteristicas = datos_mat["features"]

print("Tipo de objeto:", caracteristicas.dtype)
print("Contenido completo:")
print(caracteristicas)
```

```python
import numpy as np
import pandas as pd

# 1. Extraer los nombres de las variables
nombres_variables = [
    str(elemento[0])
    for elemento in datos_mat["features"][0]
]

print("Variables descriptivas:")
for i, nombre in enumerate(nombres_variables, start=1):
    print(f"{i}. {nombre}")

# 2. Obtener los identificadores de las ubicaciones
ids_ubicaciones = datos_mat["location_ids"].flatten()

# 3. Convertir las matrices a arreglos unidimensionales
X_entrenamiento = datos_mat["X_tr"].flatten()
X_prueba = datos_mat["X_te"].flatten()

# 4. Mostrar algunos valores para inspeccionar su estructura
print("\nPrimeros valores de X_tr:")
print(X_entrenamiento[:10])

print("\nPrimeros valores de Y_tr:")
print(datos_mat["Y_tr"][:5, :5])
```

```python
# Guardar las matrices conservando sus dimensiones
pd.DataFrame(datos_mat["Y_tr"]).to_csv(
    "Y_entrenamiento.csv", index=False
)

pd.DataFrame(datos_mat["Y_te"]).to_csv(
    "Y_prueba.csv", index=False
)

pd.DataFrame({
    "X": X_entrenamiento
}).to_csv("X_entrenamiento.csv", index=False)

pd.DataFrame({
    "X": X_prueba
}).to_csv("X_prueba.csv", index=False)

# Guardar los identificadores de las ubicaciones
pd.DataFrame({
    "location_id": ids_ubicaciones
}).to_csv("ubicaciones.csv", index=False)

print("Los archivos CSV se guardaron correctamente.")
```
